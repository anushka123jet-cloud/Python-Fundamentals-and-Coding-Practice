```python
#20. Create a basic calculator 
num1 = float(input('Enter 1st number: '))
operator = input('Enter operator (+,-,*,/): ')
num2 = float(input('Enter 2nd number: '))

if (operator == '+'):
    result = num1+num2
elif (operator == '-'):
    result = num1-num2
elif(operator == '*'):
    result = num1*num2
elif(operator == '/'):
    if(num2 != 0):
       result = num1/num2
    else:
       result = 'Cannot divide by 0'
else:
    result = 'Invalid operator'

print(result)
```

```python
#1. Check whether a string is pallindrome or not?
string = input('Enter any string: ')
reverse = string[::-1]
if(string==reverse):
    print('String is a pallindrome: ' + string)
else:
    print('Not a pallindrome')
```

```python
#2. Count the no. of vowels in a string
string = input('Enter any string: ')
count = 0
for char in string:
    if char.lower() in 'aeiou':
        count+=1
print('No. of vowels: ',count)
```

```python
#3. Count the frequency of each character 
string = input('Enter any string: ')
frequency = {}
for char in string:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1
print(frequency)
```

```python
#4. Find the first non-repeating character
string = input('Enter any string: ')
for char in string:
    if string.count(char)==1:
        print('First non-repeating character: ',char)
        break
    else:
        print('No non-repeating character found')
```

```python
#5. Remove duplicate characcters while preserving order
string = input('Enter any string: ')
result = " "
for char in string:
    if char not in result:
        result += char 
print(result)
```

```python
#6. Check whether two strings are anagrams or not.
str1 = input('Enter first string: ')
str2 = input('Enter second string: ')
if sorted(str1) == sorted(str2):
    print('Anagram')
else:
    print('Not Anagram')
```

```python
#7. Find the most frequent character 
string = input('Enter any string: ')
frequency = {}
for char in string:
    frequency[char] = frequency.get(char,0) + 1
most_frequent = max(frequency, key = frequency.get)
print('Most frequent character: ',most_frequent)
print('Frequency: ',frequency[most_frequent])
```

```python
#8. Clean name 
names = ['  ANUSHKA','Rahul  ',' priya','   Amit']
cleaned_names =[]
for name in names:
    name = name.strip()
    name = name.title()
    cleaned_names.append(name)
print(cleaned_names)
```
